```python
import numpy as np
```

```python
temperature_w1 = np.array([22.5,25.3,20.8,23.4,26.1,24.8,21.9])
print(temperature_w1)
```

```python
print("Shape:",temperature_w1.shape)
print("Data Type:",temperature_w1.dtype)
print("Number of Elements:",
      temperature_w1.size)
```

```python
fahrenheit=(temperature_w1*9/5)+32
print("Temperature in Fahrenheit:",
      fahrenheit)
```

```python
print("Maximum Temperature:",
      np.max(temperature_w1))
print("Minimum Temperature:",
      np.min(temperature_w1))
print("Mean Temperature:",
      np.mean(temperature_w1))
```

```python
print("First three days:",temperature_w1[:3])
print("Weekend temperatures:",
      temperature_w1[-2:])
print("Middle three days:",
      temperature_w1[2:5])
```

```python
print("Shape:",temperature_w1.shape)
print("Data Type:",temperature_w1.dtype)
print("Total Elements:",temperature_w1.size)
```

```python
temperatures=np.array([
    [22.5,25.3,20.8,23.4,26.1,24.8,
21.9],
     [19.2,22.5,21.3,24.0,23.5,22.8,20.1]
])
print(temperatures)
```

```python
print("Shape:",temperatures.shape)
print("Data Type:",temperatures.dtype)
print("Total Elements:",temperatures.size)
```

```python
print("First Week:",temperatures[0])
print("Second Week:",temperatures[1])
print("First Day Temperatures:",
      temperatures[:,0])
print("Lat Day Temperatures:",
      temperatures[:,-1])
```

```python
print("First Week Average:",
      np.mean(temperatures[0]))
print("Second Week Average:",
      np.mean(temperatures[1]))
print("First Week Maximum:",
      np.max(temperatures[0]))
print("Second Week Maximum:",
      np.max(temperatures[1]))
print("First Week Minimum:",
      np.min(temperatures[0]))
print("Second Week Minimum:",
      np.min(temperatures[1]))

```

```python
difference=temperatures[0]-temperatures[1]
print("Temperature Difference:")
print(difference)
print("Highest Difference:",
      np.max(np.abs(difference)))
```

```python
hottest_day=np.argmax(temperatures)
coldest_day=np.argmin(temperatures)
print("Hottest Temperature:",
      np.max(temperatures))
print("Hottest Day Index:",
      np.unravel_index(hottest_day,
        temperatures.shape))
print("Coldet Temperature:",
      np.min(temperatures))
print("Coldest Day Index:",
      np.unravel_index(coldest_day,
         temperatures.shape))                
```

```python
print("Sorted Temperatures:")
print(np.sort(temperatures))
print("All Temperatures in Ascending Order:")
print(np.sort(temperatures.flatten()))
```

```python
reshaped=temperatures.reshape(7,2)
print("Reshaped Array:")
print(reshaped)
print("New Shape:",reshaped.shape)
```

```python
transposed = temperatures.T
print("Transposed Array:")
print(transposed)
print("Transpose Shape:",transposed.shape)
```

```python
print("First Week Temperatures:")
print(temperatures[0])
print("Second Week Temperatures:")
print(temperatures[1])
print("Overall Average:")
print(np.mean(temperatures))
print("Overall Standard Deviation:")
print(np.std(temperatures))
```

```python
print("First Week Range:")
print(np.ptp(temperatures[0]))
print("Second Week Range:")
print(np.ptp(temperatures[1]))
print("Overall Median:")
print(np.median(temperatures))
print("Overall Variance:")
print(np.var(temperatures))
```

```python
first_avg = np.mean(temperatures[0])
second_avg = np.mean(temperatures[1])
print("First Week Average:",first_avg)
print("Second Week Average:",second_avg)
if first_avg > second_avg:
    print("First Week had higher average temperature")
else:
    print("Second Week had higher average temperature")
```

```python
print("First Week Highest Temperature:",
      np.max(temperatures[0]))
print("First Week Lowest Temperature:",
      np.min(temperatures[0]))
print("Second Week Highest Temperature:",
      np.max(temperatures[1]))
print("Second Week Lowests Temperature:",
      np.min(temperatures[1]))
```

```python
print("First Week Temperature:")
print(temperatures[0])
print("Second Week Temperature:")
print(temperatures[1])
print("Difference Between Weekly Averages:",
     np.mean(temperatures[0])-
np.mean(temperatures[1]))
print("Overall Average Temperature:",
      np.mean(temperatures))
```

```python
import matplotlib.pyplot as plt
days=["Day 1","Day 2","Day 3","Day 4","Day5","Day 6","Day 7"]
plt.plot(days,temperatures[0],marker='o',label="First Week")
plt.plot(days,temperatures[1],marker='o',label="Second Week")
plt.title("Weekly Temperature Comparison")
plt.xlabel("Days")
plt.ylabel("Temperature(oC)")
plt.legend()
plt.grid(True)
plt.show()

```

```python
plt.figure(figsize=(10,5))
x=np.arange(7)
width=0.35
plt.bar(x-width/2,temperatures[0],width,label="First Week")
plt.bar(x+width/2,temperatures[1],width,label="Second Week")
plt.title("Weekly Temperature Bar Chart")
plt.xlabel("Days")
plt.ylabel("Temperature(oC)")
plt.xticks(x,days)
plt.legend()
plt.grid(axis='y')
plt.show()

```

```python
plt.figure(figsize=(7,7))
average_temp=[np.mean(temperatures[0]),
              np.mean(temperatures[1])]
labels=["First Week","Second Week"]
plt.pie(average_temp,labels=labels,
        autopct='%1.1f%%',
        startangle=90)
plt.title("Weekly Average Temperature Distribution")
plt.show()
              
```

```python
plt.figure(figsize=(8,5))
plt.hist(temperatures.flatten(),bins=7,edgecolor='black')
plt.title("Temperature Distribution")
plt.xlabel("Temperature(oC)")
plt.ylabel("Frequency")
plt.grid(axis='y')
plt.show()
```

```python
import matplotlib.pyplot as plt
print(plt)
print(hasattr(plt,'boxplot'))
```

```python
fig,ax=plt.subplots(figsize=(8,5))
ax.boxplot([temperatures[0],temperatures[1]],
            tick_labels=["First Week","Second Week"])
ax.set_title("Weekly Temperature Box Plot")
ax.set_xlabel("Weeks")
ax.set_tick_abel("Temperature(oC)")
ax.grid(True)
plt.show()
```

```python
print("Temperature Data Analysis")
print("First Week Average:",
      np.mean(temperatures[0]))
print("Second Week Average:",
      np.mean(temperatures[1]))
print("First Week Median:",
      np.median(temperatures[0]))
print("Second Week Median:",
      np.median(temperatures[1]))
print("Conclusion:")
print("First Week had higher average temperature.")
```
